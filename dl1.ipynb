Task 1: Vectorized Multi-Layer Perceptron (MLP) Forward & Backward
Pass from Scratch


```python
#Import Libraries
import numpy as np
np.random.seed(42)
```

```python
#Create Dummy Dataset
# Number of samples
m = 200

# Features
X = np.random.randn(m, 2)

# Labels
y = np.random.randint(0, 2, size=(m,))

print(X.shape)
print(y.shape)
```

```python
#One Hot Encoding
def one_hot(y, num_classes):

    onehot = np.zeros((len(y), num_classes))
    onehot[np.arange(len(y)), y] = 1

    return onehot

Y = one_hot(y,2)
```

```python
#Activation Functions
def relu(z):
    return np.maximum(0,z)

def relu_derivative(z):
    return (z>0).astype(float)
```

```python
#Softmax
def softmax(z):

    exp = np.exp(z - np.max(z,axis=1,keepdims=True))

    return exp / np.sum(exp,axis=1,keepdims=True)
```

```python
#Cross Entropy Loss
def cross_entropy(y_true,y_pred):

    m = y_true.shape[0]

    loss = -np.sum(y_true*np.log(y_pred+1e-10))/m

    return loss
```

```python
#Initialize Parameters
input_size = 2
hidden1 = 8
hidden2 = 6
output_size = 2

W1 = np.random.randn(input_size,hidden1)*0.01
b1 = np.zeros((1,hidden1))

W2 = np.random.randn(hidden1,hidden2)*0.01
b2 = np.zeros((1,hidden2))

W3 = np.random.randn(hidden2,output_size)*0.01
b3 = np.zeros((1,output_size))
```

```python
#Forward Pass
def forward(X):

    global Z1,A1,Z2,A2,Z3,A3

    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = relu(Z2)

    Z3 = A2 @ W3 + b3
    A3 = softmax(Z3)

    return A3
```

```python
#Backward Pass
def backward(X,Y):

    global W1,W2,W3,b1,b2,b3

    m = X.shape[0]

    dZ3 = A3 - Y

    dW3 = A2.T @ dZ3 / m
    db3 = np.sum(dZ3,axis=0,keepdims=True)/m

    dA2 = dZ3 @ W3.T
    dZ2 = dA2 * relu_derivative(Z2)

    dW2 = A1.T @ dZ2 / m
    db2 = np.sum(dZ2,axis=0,keepdims=True)/m

    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * relu_derivative(Z1)

    dW1 = X.T @ dZ1 / m
    db1 = np.sum(dZ1,axis=0,keepdims=True)/m

    lr = 0.1

    W3 -= lr*dW3
    b3 -= lr*db3

    W2 -= lr*dW2
    b2 -= lr*db2

    W1 -= lr*dW1
    b1 -= lr*db1
```

```python
#Training
epochs = 500

for epoch in range(epochs):

    pred = forward(X)

    loss = cross_entropy(Y,pred)

    backward(X,Y)

    if epoch%50==0:
        print(f"Epoch {epoch} Loss : {loss:.4f}")
```

```python
#Prediction
pred = forward(X)

predictions = np.argmax(pred,axis=1)

accuracy = np.mean(predictions==y)

print("Accuracy:",accuracy)
```
